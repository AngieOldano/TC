## Ejercicio 1: 
Calcular la imagen especular de una cadena definida sobre {a, b}, es decir, f(w)=reverso(w). Ejemplos: f(aabb)=bbaa y f(aba)=aba

### Definición de la MT
#### M = ⟨{a, b, x, y, X, Y, □}, {a, b}, □, {q0, q1, q2, q3, q4, q5}, q0, {q5}, δ⟩

**Alfabeto de cinta:** Γ = {a, b, x, y, X, Y, □}

**Alfabeto de entrada:** Σ = {a, b}

**Blanco:** □

**Estados:** Q = {q0, q1, q2, q3, q4, q5}

**Estado inicial:** q0

**Estados finales:** F = {q5}

**Función de transición:** δ: Q x Γ → Q x Γ x {L, R, S}


### Tabla de transiciones

| Q   | a          | b          | x          | y          | X          | Y          | □          |
|-----|------------|------------|------------|------------|------------|------------|------------|
| >q0 | (q0, a, R) | (q0, b, R) | -          | -          | -          | -          | (q1, □, L) |
| q1  | (q2, x, R) | (q3, y, R) | (q1, x, L) | (q1, y, L) | (q1, X, L) | (q1, Y, L) | (q4, □, R) |
| q2  | (q2, a, R) | (q2, b, R) | (q2, x, R) | (q2, y, R) | (q2, X, R) | (q2, Y, R) | (q1, X, L) |
| q3  | (q3, a, R) | (q3, b, R) | (q3, x, R) | (q3, y, R) | (q3, X, R) | (q3, Y, R) | (q1, Y, L) |
| q4  | -          | -          | (q4, □, R) | (q4, □, R) | (q4, a, R) | (q4, b, R) | (q5, □, S) |
| *q5 | -          | -          | -          | -          | -          | -          | -          |


![alt text](1-reverse/1-reverse.png)

### Funcionamiento

- **q0:** avanza sobre toda la palabra de entrada sin modificarla, hasta llegar al blanco de la derecha. Ahí retrocede un paso para posicionarse sobre el último símbolo.

- **q1:** marca el símbolo actual con x o y (dependiendo de si es a o b) y avanza hacia la derecha. Tambien retrocede los simbolos ya marcados. Si encuentra un blanco en lugar de un símbolo marcado, pasa a q4.

- **q2:** rescribe el blanco con X que corresponde a la a marcada en q1. Tambien avanza sobre los símbolos marcados y no marcados.

- **q3:** rescribe el blanco con Y que corresponde a la b marcada en q1. Tambien avanza sobre los símbolos marcados y no marcados.

- **q4:** una vez que esten todos los simbos marcados y la palabra duplicada esté en la cinta, recorre la cinta de izquierda a derecha, restaurando los símbolos: las marcas mayúsculas (X, Y) vuelven al símbolo correspondiente (a, b), y las minúsculas (x, y) se reescribien por blanco.

- **q5:** estado de aceptación.

## Ejercicio 2:
- Duplicar una cadena de aes y bes en la cinta. Ejemplo: si la MT comienza con abbaa□ en su cinta, luego de procesar su programa debe terminar con abbaa□abbaa

### Definición de la MT

#### M = ⟨{a, b, x, y, z, □}, {a, b}, □, {q4, q5, p, q1, q2, q3, q7, q8}, q4, {q8}, δ⟩

**Alfabeto de entrada:** Σ = {a, b}

**Alfabeto de cinta:** Γ = {a, b, x, y, z, □}

**Blanco:** □

**Estados:** Q = {q4, q5, p, q1, q2, q3, q7, q8}

**Estado inicial:** q4

**Estados finales:** F = {q8}

**Función de transición:** δ: Q x Γ → Q x Γ x {L, R, S}

### Tabla de transiciones

| Q | a | b | x | y | z | □ |
|---|---|---|---|---|---|---|
| >q4 | (q4, a, R) | (q4, b, R) | - | - | - | (q5, z, L) |
| q5 | (q5, a, L) | (q5, b, L) | - | - | - | (p, □, R) |
| p | (q1, x, R) | (q3, y, R) | - | - | (q7, □, L) | - |
| q1 | (q1, a, R) | (q1, b, R) | - | - | (q1, z, R) | (q2, a, L) |
| q2 | (q2, a, L) | (q2, b, L) | (p, x, R) | (p, y, R) | (q2, z, L) | - |
| q3 | (q3, a, R) | (q3, b, R) | - | - | (q3, z, R) | (q2, b, L) |

| *q8 | - | - | - | - | - | - |

![alt text](2-duplicar/2-duplicar.png)

### Funcionamiento

- **q4:** avanza sobre toda la palabra de entrada sin modificarla. Al llegar al primer blanco, lo reemplaza por z (para distinguirlo de los demas blancos) y retrocede ubicandose en el ultimo simbolo de la palabra de entrada.

- **q5:** retrocede sobre toda la entrada y se ubica en el primer símbolo de la entrada

- **p:** lee el primer símbolo sin marcar de la entrada. Si es a, lo marca como x y va a q1; si es b, lo marca como y y va a q3. Si en cambio lee z, significa que ya se marcaron todos los símbolos y reescribe la z por blanco como estaba originalmente.

- **q1:** realiza la copia de a : avanza hacia la derecha salteando el resto de la entrada y la marca z, hasta encontrar el primer blanco al final de la cinta y lo reescribe por a.

- **q2:** retrocede saltando la z y los símbolos ya copiados y no copiados, hasta encontrar la marca (x o y) dejada por p. La respeta sin modificarla y avanza  hacia adelante, volviendo a p para procesar el siguiente símbolo.

- **q3:** realiza la copia de b : avanza hacia la derecha salteando el resto de la entrada y la marca z, hasta encontrar el primer blanco al final de la cinta y lo reescribe por b.

- **q7:** retrocede sobre la entrada completamente marcada, reescribiendo cada marca de vuelta a su símbolo original (x → a, y → b), hasta llegar al blanco inicial.

- **q8:** estado de aceptación, sin transiciones salientes.

## Ejercicio3:
Se dispone de una cinta en la que hay un número m de 1s seguido de un número n ≥ m de Aes. Se desea definir una MT que cambie las primeras m Aes por Bes. Se supone que la cabeza de la cinta inicialmente está en el 1 más a la izquierda

### Definición de la MT

#### M = ⟨{a, b, 1, x, □}, {a, b, 1}, □, {q0, q1, q2, q3, q4, q5}, q0, {q5}, δ⟩

**Alfabeto de entrada:** Σ = {a, b, 1}

**Alfabeto de cinta:** Γ = {a, b, 1, x, □}

**Blanco:** □

**Estados:** Q = {q0, q1, q2, q3, q4, q5}

**Estado inicial:** q0

**Estados finales:** F = {q5}

**Función de transición:** δ: Q x Γ → Q x Γ x {L, R, S}

### Tabla de transiciones

| Q | a | b | 1 | x | □ |
|---|---|---|---|---|---|
| >q0 | (q3, a, L) | (q3, b, L) | (q1, x, R) | (q0, x, R) | - |
| q1 | (q2, b, L) | (q1, b, R) | (q1, 1, R) | - | - |
| q2 | - | (q2, b, L) | (q2, 1, L) | (q2, x, L) | (q0, □, R) |
| q3 | - | (q3, b, L) | (q3, 1, L) | (q3, x, L) | (q4, □, R) |
| q4 | (q5, a, S) | (q5, b, S) | - | (q4, 1, R) | - |
| *q5 | - | - | - | - | - |

### Funcionamiento
- **q0:** avanza hacia la derecha saltando los 1 ya procesados (marcados como x). Cuando encuentra un 1 lo marca como x y pasa a q1 a buscar una a para sobreescribir. Si en cambio encuentra directamente a o b (ya no quedan 1 sin procesar), retrocede un paso y pasa a q3 para terminar.

- **q1:** avanza hacia la derecha saltando b y 1 sin procesar, buscando la primera a. Al encontrarla, la convierte en b y retrocede, pasando a q2.

- **q2:** retrocede saltando x, b y 1 hasta llegar al blanco del inicio de la cinta. Ahí avanza un paso y vuelve a q0, para procesar el siguiente 1.

- **q3:** con todos los 1 ya marcardos retrocede hasta el blanco inicial. Al llegar al blanco, avanza un paso posiciononandose sobre el primer símbolo y pasa a q4.

- **q4:** avanza hacia la derecha restaurando cada x a su valor original 1 hasta encontrar a o b y pasa a q5.

- **q5:** estado de final de aceptación.

### Funcionamiento
#### Estrategia:
Se dividio el problema en tres partes:
1. Duplicar una cadena de 1s.
2. Repetir la duplicacion por la cantidad de 1s - 1 (-1 porque el primer 1 no se duplica ya es la cadena original) que hay a la izquierda del * . Por ejemplo, si la entrada es 111 * 1111, se duplica la cadena de 1s a la derecha del * dos veces.
3. El ultimo paso es encontrar la mitad de la cadena duplicada y eliminarla, para obtener el resultado de la division por dos. Si el resultado es impar, se elimina el 1 sobrante.

##### 1. Duplicar una cadena de 1s.
- q0: marca el primer 1 a la izquierda de todo con z y avanza a q1. Cuando no quedan 1s sin marcar, pasa a q3.
- q1: avanza hasta encontrar el primer blanco a la derecha de la cadena de 1s y lo reemplaza por x, luego retrocede a q2
- q2: retrocede hasta encontrar el primer z y lo respeta, luego avanza a q0 para repetir el proceso hasta que no queden 1s sin marcar.
- q3: reemplaza todos los x (que representan la cadena ya duplicada)por 1 y retrocede a q4.
- q4: conecta con la parte para duplicar la cadena: retrocede recorriendo los 1s y zs hasta encontrar el *, lo respeta y avanza a q8 .

![alt text](<6-multiplicar dos numeros y dividir por dos/duplicar cadena.png>)

##### 2. Repetir la duplicacion Duplicar una cadena de 1s.
- q5: elimina el primer 1 reescribiendolo por blanco y avanza a q6.
- q6: marca el 1 por z y avanza a q7. Cuando no quedan 1s sin marcar, pasa a q9.
- q7: avanza por todos los 1s de la izquierda y el *. Va a q0 para pasar por la **duplicacion de la cadena de 1s**.
- q8: despues de duplicar la cadena, retrocede por toda la cadena de la izquierda (por los 1s no marcados y las zs) para ir a q6 y repetir el proceso.
- q9: vuelve al primer blanco pasando por todos las zs y se ubica en la primera z.
- q10: borra todas las zs de la izquierda y el * reemplazandolos por blancos para que quede en la cinta la cadena resultante de la multiplicacion y avanza a q11.
- q11: reemplaza las zs que qudaron a la derecha del * por 1s y avanza a q12 pasando por un uno.
- q12: avanza por los 1s restantes hasta encontrar el primer blanco y pasa a q13 posicionandose en el ultimo simbolo de la cadena.

![alt text](<6-multiplicar dos numeros y dividir por dos/repetir duplicacion.png>)

##### 3. Dividir la cadena a la mitad.
- q13: marca el ultimo 1 de la cadena con x y retrocede a q14 para ir a buscar el primer 1 de la cadena. Cuando no quedan 1s sin marcar el cabezal esta en  **la mitad de la cadena** entonces pasa a q17 para eliminar la mitad de la cadena reescribiendo con blanco la x.
- q14: retrocede por todos los 1s hasta encontrar el primer blanco y pasa a q15 posicionandose en el primer simbolo de la cadena.
- q15: marca el primer 1 de la cadena con x y avanza a q16. Cuando no quedan 1s sin marcar el cabezal esta en  **la mitad de la cadena** entonces pasa a q17 para eliminar la mitad de la cadena reescribiendo con blanco la x.
- q16: avanza por todos los 1s entre las xs hasta encontrar el ultimo 1 no marcado de la cadena posicionandose en el. Avanza a q15 para repetir el proceso hasta que no queden 1s sin marcar.
- q17: borra toda la mitad de la cadena reescribiendo con blanco las xs y avanza a q18.
- q18: avanza por todos los blancos hasta encontrar el primer simbolo de la cadena y pasa a q21.
- q21: reescribe todos los xs por 1s y avanza a q20.
- q20: estado de aceptación.
