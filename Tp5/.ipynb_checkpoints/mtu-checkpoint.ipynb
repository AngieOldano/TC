```python
1. Definición de la MT M

M = ⟨{0, 1, □}, {0, 1}, □, {q0, q1, q2}, q0, {q2}, δ⟩

  Q	 |      0	  |      1	   |      □
-----|------------|------------|---------------
>q0	 | (q0, 0, R) |	(q1, 1, R) | (q2, □, R)
q1	 | (q1, 0, R) |	(q0, 1, R) |     -
*q2	 |      -	  |       -	   |     -

2. Codificación de elementos
Símbolos:
0 = 00, 1 = 01, □ = 10
Estados:
q0 = 00, q1 = 01, q2 = 10
Movimientos:
L = 1, R = 0

3. Codificación de transiciones
  Q	 |      00	   |      01	 |      10
-----|-------------|-------------|---------------
>00  | (00, 00, 0) | (01, 01, 0) | (10, 10, 0)
01	 | (01, 00, 0) | (00, 01, 0) |     -
*10	 |      -	   |       -	 |     -


4. Entrada para la MTU

donde:
⟨M,w⟩
⟨M⟩ es la lista de transiciones codificadas
w es la cadena de entrada, por ejemplo 001100

Ejemplo de entrada codificada:
*000101000010$0000#000000000#000101010#001010100#010001000#010100010

```
